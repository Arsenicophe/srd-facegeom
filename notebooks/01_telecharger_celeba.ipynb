Chaque fichier est vérifié par son empreinte MD5 officielle (celle de

Relancer ne retélécharge que ce qui manque ou est faux. Pas besoin de GPU ici :
un serveur CPU suffit.

## 1. Drive et code

```python
from google.colab import drive
drive.mount("/content/drive")

import os, subprocess
from pathlib import Path

REPO_URL = "https://github.com/Arsenicophe/srd-facegeom.git"
CODE_DIR = Path("/content/srd-facegeom")
if CODE_DIR.exists():
    subprocess.run(["git", "-C", str(CODE_DIR), "pull", "--ff-only"], check=True)
else:
    subprocess.run(["git", "clone", REPO_URL, str(CODE_DIR)], check=True)
os.chdir(CODE_DIR)
print(subprocess.run(["git", "log", "-1", "--oneline"], capture_output=True, text=True).stdout)
```
